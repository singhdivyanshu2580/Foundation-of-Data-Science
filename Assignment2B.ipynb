```python
import pandas as pd

# Load Dataset
df = pd.read_csv("employee_dataset.csv")
print(df.head())
```

```python
df.isnull().sum()           #check missing values
#df.fillna(df.mean(), inplace=True)  #fill missing values with mean# fill NaN with Mean
df.dropna()     # drop rows with NaN values
```

```python
df.isnull().sum()   # check missing values again
```

```python
df.fillna(0, inplace=True)  # fill NaN with 0
```

```python
# df.fillna(df.mean(), inplace=True) #fill missing values with mean
# df.dropna() # drop rows with NaN values
```

```python
df.duplicated().sum()   # check duplicate values
```

```python
df.drop_duplicates(inplace=True)    # remove duplicates
```

```python
print(df.head())
```

```python
print(df.groupby('Department')[['Age', 'Salary', 'PerformanceScore', 'WorkHours']].mean())  # Group by Department and calculate mean of numerical columns
```

```python
df.groupby('Department')['Salary'].agg(["mean", "min", "max", "std", 'count'])  # Group by Department and calculate mean, min, max, std, count of Salary
```

```python
df.groupby(["Department",'Age'])['Salary'].mean()
```
